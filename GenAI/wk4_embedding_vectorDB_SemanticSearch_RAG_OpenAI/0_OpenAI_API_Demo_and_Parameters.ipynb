# Temperature
- Controls Creativity

```python
# Create a chat completion request

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a fantasy story writer."},
        {"role": "user", "content": "Suggest a plot for a short fantasy story for kids."}
    ],
    temperature=0.1
)

# Print the model's response
print("AI Response:", response.choices[0].message.content)
```

# Max tokens

```python
messages = [
    {"role": "system", "content": "You are a fantasy story writer."},
    {"role": "user", "content": "Write a detailed, three-paragraph fantasy story plot about a dragon who lost his fire."}
]

# Test 1: Aggressive limit (Forces truncation mid-sentence)
truncated_res = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages,
    temperature=0.7,
    max_tokens=30
)

# Test 2: Generous limit (Allows completion)
completed_res = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages,
    temperature=0.7,
    max_tokens=300
)

# Extract choice data to view finish reasons
choice_1 = truncated_res.choices[0]
choice_2 = completed_res.choices[0]

print(f"=== Test 1: max_tokens=30 ===")
print(f"Finish Reason: '{choice_1.finish_reason}'")
print(f"Output:\n{choice_1.message.content}\n")

print(f"=== Test 2: max_tokens=300 ===")
print(f"Finish Reason: '{choice_2.finish_reason}'")
print(f"Output:\n{choice_2.message.content}\n")
```

# n: 
Determines how many independent chat completion choices to generate for a single prompt. For instance, setting n=3 generates 3 entirely separate alternative responses in your response.choices array (note: this will triple your generation costs for that specific request).

```python
# Create a chat completion request

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a fantasy story writer."},
        {"role": "user", "content": "Suggest a plot for a short fantasy story in less than 500 words for kids."}
    ],
    temperature=0.8,
    n=2
)

# Loop through and print both generated choices
for index, choice in enumerate(response.choices):
    print(f"\n=== AI Response Variation {index + 1} ===")
    print(choice.message.content)
```

# top_p (Nucleus Sampling): 
An alternative to temperature that controls response diversity by restricting the pool of considered words to a cumulative probability percentage.

For example, top_p=0.9 means the model only selects from the top 90% most likely words. 


## Rule of thumb: Adjust either temperature or top_p, but never both at the same time.

```python
# Create a chat completion request

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a fantasy story writer."},
        {"role": "user", "content": "Suggest a plot for a short fantasy story in less than 500 words for kids."}
    ],
    top_p=0.9,
    n=1
)

# Print the model's response
print("AI Response:", response.choices[0].message.content)
```

# Output: JSON format

```python

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system", 
            "content": "You are a customer support triage bot. You must extract key details and return them ONLY as a valid JSON object."
        },
        {
            "role": "user", 
            "content": (
                "Hey, I bought a MacBook Pro (Serial: PRO-99281) yesterday and the screen "
                "is flickering. My email is alex@email.com and I need a replacement ASAP! "
                "Order ID #4402."
            )
        }
    ],
    # Enforces that the model outputs a valid JSON string
    response_format={"type": "json_object"},
    temperature=0.2
)

# Print the clean JSON string
print(response.choices[0].message.content)
```

# frequency_penalty
A number between -2.0 and 2.0. 

Positive values penalize words based on how many times they have already appeared in the text, lowering the likelihood that the model repeats the exact same phrases or verbatim expressions.

```python
system_instruction = (
    "You are an enthusiastic gadget reviewer. Write a detailed, "
    "long-form marketing review for a new smartphone."
)

user_prompt = (
    "Write a 150-word review focusing on the camera quality, "
    "battery life, and overall design."
)

# Test: Applying a frequency penalty to force diverse wording
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_instruction},
        {"role": "user", "content": user_prompt}
    ],
    temperature=0.7,
    
    # High penalty forces the AI to actively avoid repeating any words/phrases
    frequency_penalty=1.2 
)

print("=== Diverse Marketing Copy ===")
print(response.choices[0].message.content)
```

# presence_penalty
A number between -2.0 and 2.0. 

Positive values penalize the model for repeating words or topics it has already used, actively forcing it to introduce new subject matter.

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system", 
            "content": "You are a startup innovation consultant."
        },
        {
            "role": "user", 
            "content": "Brainstorm 5 radically different features for a new fitness app."
        }
    ],
    temperature=0.8,
    
    # High presence penalty forces the model to pivot to completely new concepts
    # for each bullet point, rather than expanding on the same theme.
    presence_penalty=1.5 
)

print("=== Diversified Feature Concepts ===")
print(response.choices[0].message.content)
```
