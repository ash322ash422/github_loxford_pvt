# Using Pre-trained Models
Gensim provides a built-in downloader called gensim.downloader to easily load popular pre-trained vectors.

```python
!pip install gensim==4.4.0
```

```python
import gensim.downloader as api

# Load a lightweight pre-trained model (Twitter vectors, 25 dimensions)
# For standard Google News 300d vectors, use: api.load('word2vec-google-news-300')
print("Downloading/Loading pre-trained embedding model...")

model = api.load("glove-twitter-25") # Took 3 minutes
```

## Lets check where this file is saved and also the size

```python
import os

# 1. Get the local file path where Gensim caches the model
# Gensim stores metadata about where files are saved in its internal api registry
model_info = api.info("glove-twitter-25")
cache_dir = os.path.expanduser("~/gensim-data/glove-twitter-25")
file_name = f"glove-twitter-25.gz"
full_path = os.path.join(cache_dir, file_name)

# 2. Print the location
print(f"\nModel Local Location: {full_path}")

# 3. Calculate and print the download file size from disk
if os.path.exists(full_path):
    file_size_bytes = os.path.getsize(full_path)
    file_size_mb = file_size_bytes / (1024 * 1024)
    print(f"Downloaded File Size: {file_size_mb:.2f} MB")
else:
    print(
        "Could not auto-detect file on disk. Default path may vary on this system."
    )

# 4. Check the approximate memory footprint (RAM) of the loaded vectors
memory_bytes = model.vectors.nbytes
memory_mb = memory_bytes / (1024 * 1024)
print(f"Loaded Vector Matrix Size in RAM: {memory_mb:.2f} MB")
```

## Find vectors and similarities

```python
vec = model["computer"]

print("Vector representation of 'computer':", vec)
```

```python
result = model.most_similar("computer")

print("\nMost similar words to 'computer':")
for word, similarity in result[:7]:
    print(f"{word}: {similarity:.4f}")
```

```python
vec = model["water"]
print("Vector representation of 'water':", vec)
```

```python
result = model.most_similar("water")

print("\nMost similar words to 'water':")
for word, similarity in result[:7]:
    print(f"{word}: {similarity:.4f}")
```

# OPTIONAL (Delete the file to save space)

```python
import os
import shutil

# Locate the specific model folder in the default gensim-data directory
cache_dir = os.path.expanduser("~/gensim-data/glove-twitter-25")

if os.path.exists(cache_dir):
    # Remove the directory and all its contents
    shutil.rmtree(cache_dir)
    print(f"Successfully deleted model files from: {cache_dir}")
else:
    print(f"No files found at {cache_dir}. It may have already been deleted.")
```
