# **Generating Images with OpenAI Image Models and the OpenAI API**

### **Objective:**  
To generate images from textual descriptions using **OpenAI’s gpt-image-2** model, demonstrating how AI can convert creative text prompts into stunning visuals.


---

### **Steps to perform:**
1. Set up the environment  
2. Generate an image from text  
3. Display the generated image  

---






### **Step 1: Set up the environment**
- Import the required libraries
- Initialize the **OpenAI client** for making API calls to the gpt-image-1 model



```python
# Import libraries
from openai import OpenAI
import base64
import matplotlib.pyplot as plt
from PIL import Image
from io import BytesIO

# Initialize OpenAI client
client = OpenAI()

print("✅ Environment setup complete.")
```

### **Step 2:  Generate image from the text**
- Define a text prompt describing the image you want to generate
- Use the OpenAI Images API to generate an image using the "gpt-image-1" model. 
- Decode and save the image locally

```python
# Define the text prompt
text_prompt = input("Enter a text prompt to generate an image: ")

# Generate an image using OpenAI's gpt-image-1
response = client.images.generate(
    model="gpt-image-2",
    prompt=text_prompt,
    size="1024x1024"
)

# Extract base64-encoded image data
image_base64 = response.data[0].b64_json

# Decode the base64 image
image_bytes = base64.b64decode(image_base64)

# Save the image as a PNG file
output_path = "generated_image.png"
with open(output_path, "wb") as f:
    f.write(image_bytes)

print(f"✅ Image successfully generated and saved as '{output_path}'.")
```

### __Step 3: Display the generated image__
- Load the generated image using Pillow (PIL)
- Display it inline using Matplotlib for visual confirmation

```python
# Load and display the generated image
image = Image.open(BytesIO(image_bytes))

plt.figure(figsize=(6, 6))
plt.imshow(image)
plt.axis("off")
plt.title(f"Prompt: {text_prompt}", fontsize=12)
plt.show()

print("✅ Image displayed successfully.")
```

**Observation:**
- The image will be displayed inline in the notebook and also saved locally as generated_image.png.
- You can open it manually from your project folder if needed.
