```python
import importlib.metadata

# List the distribution package names
packages = ["openai"]

for package in packages:
    try:
        version = importlib.metadata.version(package)
        print(f"{package} version: {version}")
    except importlib.metadata.PackageNotFoundError:
        print(f"{package} is not installed in this environment.")
```

```python
from openai import OpenAI

# import os
# from dotenv import load_dotenv
# load_dotenv()
# endpoint = os.getenv("AZURE_OPENAI_ENDPOINT")
# model_deployment_name = os.getenv("AZURE_OPENAI_DEPLOYMENT_NAME")
# api_key = os.getenv("AZURE_OPENAI_API_KEY")


endpoint = "https://intst-azure-openai.services.ai.azure.com/openai/v1"
deployment_name = "gpt-5.6-sol" # or gpt-5.1 which is cheaper
api_key = "key_here"

client = OpenAI(
    base_url=endpoint,
    api_key=api_key
)

response = client.responses.create(
    model=deployment_name,
    input="What is the capital of France?",
)

print(f"answer: {response.output[0]}")
```

```python
# Extract the first item in the content list and read its 'text' attribute
print(f"text: {response.output[0].content[0].text}")
```
