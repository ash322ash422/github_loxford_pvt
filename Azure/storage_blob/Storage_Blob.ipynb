```python
import os
from dotenv import load_dotenv

# Load the keys from your local .env file into Python's environment memory
load_dotenv()

CONNECTION_STRING = os.getenv("AZURE_STORAGE_CONNECTION_STRING")
```

```python
import os
from azure.storage.blob import BlobServiceClient
from azure.core.exceptions import ResourceExistsError

# 1. Initialize the client using your Connection String
blob_service_client = BlobServiceClient.from_connection_string(CONNECTION_STRING)

# 2. Programmatically create a new container
container_name = "my-automation-container"

try:
    container_client = blob_service_client.create_container(container_name)
    print(f"Container '{container_name}' created successfully.")
except ResourceExistsError:
    print(f"Container '{container_name}' already exists. Proceeding...")
    container_client = blob_service_client.get_container_client(container_name)

```

```python
# 3. Upload a file (Blob) to the container

content = "Weather is nice !!"

blob_name = "sample.txt"
blob_client = container_client.get_blob_client(blob_name)

print(f"Uploading {blob_name} to blob storage...")
blob_client.upload_blob(content, overwrite=True)
print("Upload complete!")
```

```python
# 4. List all blobs in the container
print("\nListing blobs in container:")
blob_list = container_client.list_blobs()
for blob in blob_list:
    print(f" - {blob.name}")
```

```python
# download t file
blob_client = container_client.get_blob_client("sample.txt")

with open("downloaded-file.txt", "wb") as download_file:
    download_file.write(blob_client.download_blob().readall())
```

# Load an image

```python
from azure.storage.blob import ContentSettings


#  Define your paths and metadata
container_name = "my-automation-container"
local_image_path = "sample_picture.jpg"        # Path to the image on your computer
blob_name = "uploaded_images/photo.jpg"  # Target path inside Azure (creates a virtual folder)

container_client = blob_service_client.get_container_client(container_name)
blob_client = container_client.get_blob_client(blob_name)

# 3. Define the correct image MIME type
# Use image/png for PNG files, image/gif for GIFs, etc.
image_settings = ContentSettings(content_type="image/jpeg")

# 4. Open the file in binary mode and upload
print(f"Uploading '{local_image_path}' to Azure Storage...")

with open(local_image_path, "rb") as image_file:
    blob_client.upload_blob(
        image_file, 
        overwrite=True, 
        content_settings=image_settings
    )

print(f"Successfully uploaded! The image is saved as: {blob_name}")
```
