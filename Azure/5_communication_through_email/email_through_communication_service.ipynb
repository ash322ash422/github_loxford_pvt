```python
from azure.communication.email import EmailClient

import os
from dotenv import load_dotenv

load_dotenv()

EMAIL_CONNECTION_STRING = os.getenv("AZURE_EMAIL_CONNECTION_STRING")
SENDER = os.getenv("AZURE_EMAIL_SENDER")

def send_email_through_azure( recipient, subject, body):

    client = EmailClient.from_connection_string(EMAIL_CONNECTION_STRING)

    message = {
        "senderAddress": SENDER,
        "recipients": {
            "to": [
                {
                    "address": recipient
                }
            ]
        },

        "content": {
            "subject": subject,
            "plainText": body
        }

    }

    poller = client.begin_send(message)

    result = poller.result()

    print(result)
```

```python
RECIPIENT = "ash322.ash422@gmail.com"
SUBJECT="Demo of Azure Email Service"
BODY="Congratulations!!!!"

send_email_through_azure( recipient=RECIPIENT,
            subject=SUBJECT,
            body=BODY
)
```
