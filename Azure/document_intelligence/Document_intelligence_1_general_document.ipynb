```python
import os
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

endpoint = os.getenv("FORM_RECOGNIZER_ENDPOINT")
key = os.getenv("FORM_RECOGNIZER_KEY")
```

```python
"""
This code sample shows Prebuilt Document operations with the Azure Form Recognizer client library. 
The async versions of the samples require Python 3.6 or later.

To learn more, please visit the documentation - Quickstart: Form Recognizer Python client library SDKs
https://learn.microsoft.com/azure/applied-ai-services/form-recognizer/quickstarts/get-started-v3-sdk-rest-api?view=doc-intel-3.1.0&pivots=programming-language-python
"""

from azure.core.credentials import AzureKeyCredential
from azure.ai.formrecognizer import DocumentAnalysisClient

"""
Remember to remove the key from your code when you're done, and never post it publicly. For production, use
secure methods to store and access your credentials. For more information, see 
https://docs.microsoft.com/en-us/azure/cognitive-services/cognitive-services-security?tabs=command-line%2Ccsharp#environment-variables-and-application-configuration
"""
# endpoint = "YOUR_FORM_RECOGNIZER_ENDPOINT"
# key = "YOUR_FORM_RECOGNIZER_KEY"

# sample document
formUrl = "https://raw.githubusercontent.com/Azure-Samples/cognitive-services-REST-api-samples/master/curl/form-recognizer/sample-layout.pdf"
# formUrl = "https://raw.githubusercontent.com/ash322ash422/data/main/pdf/sample_layout_microsoft.pdf"

document_analysis_client = DocumentAnalysisClient(
    endpoint=endpoint, 
    credential=AzureKeyCredential(key)
    )
    
poller = document_analysis_client.begin_analyze_document_from_url("prebuilt-document", formUrl)
result = poller.result()

print("----Key-value pairs found in document----")
for kv_pair in result.key_value_pairs:
    if kv_pair.key and kv_pair.value:
        print("Key '{}': Value: '{}'".format(kv_pair.key.content, kv_pair.value.content))
    else:
        print("Key '{}': Value:".format(kv_pair.key.content))

print("----------------------------------------")
```

# 1. Extracted Text Blocks (Paragraphs)

It groups text into semantic paragraphs and identifies their role (like headings, section titles, or page numbers).

```python
print("----Paragraphs----")
for paragraph in result.paragraphs:
    # role can be 'heading', 'title', 'sectionHeading', etc.
    role_str = f" [{paragraph.role}]" if paragraph.role else ""
    print(f"Paragraph{role_str}: {paragraph.content}")
```

# 2. Full Layout Tables

If your document contains tables, it reconstructs the grid layout, detailing rows, columns, and spanning cells.

```python
print("----Tables----")
for table_idx, table in enumerate(result.tables):
    print(f"Table # {table_idx} has {table.row_count} rows and {table.column_count} columns")
    for cell in table.cells:
        print(f"...Cell[{cell.row_index}][{cell.column_index}] text: '{cell.content}'")
```

# 3. Read Layout (Pages, Lines, and Words)

You can break the document down page-by-page to find lines of text, specific words, and their exact visual coordinates (bounding boxes) on the page.

```python
print("----Pages and Lines----")
for page in result.pages:
    print(f"Page number: {page.page_number} (Width: {page.width}, Height: {page.height} {page.unit})")
    for line_idx, line in enumerate(page.lines):
        print(f"...Line # {line_idx} text: '{line.content}'")
        # You can also access page.words for individual word tokens
```

# 4. Selection Marks (Checkboxes and Radio Buttons)

It automatically detects checkboxes and tells you if they are filled (selected) or empty (unselected).

```python
print("----Selection Marks----")
for page in result.pages:
    for mark in page.selection_marks:
        print(f"Selection Mark is '{mark.state}' at bounding box {mark.polygon}")
```

# 5. Document Styles (Handwritten Text Detection)

It identifies whether specific sections of text were typed or handwritten.

```python
print("----Styles (Handwriting Detection)----")
for style in result.styles:
    if style.is_handwritten:
        print("Found handwritten content:")
        for span in style.spans:
            # Extract the actual handwritten text using the span offset
            print(f"...Handwritten text: '{result.content[span.offset : span.offset + span.length]}'")
```
