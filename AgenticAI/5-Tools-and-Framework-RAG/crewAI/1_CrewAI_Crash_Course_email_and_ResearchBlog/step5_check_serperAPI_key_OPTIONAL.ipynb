```python
import os
import requests
from dotenv import load_dotenv

# 1. Load your .env file if the key is stored there
load_dotenv()

# 2. Set your key directly or fetch it from environment variables
# Replace 'YOUR_ACTUAL_KEY' with your string if it's not in your .env file
SERPER_API_KEY = os.getenv("SERPER_API_KEY") or "YOUR_ACTUAL_KEY"


# Ensure the environment variable is globally visible to CrewAI tools later
os.environ["SERPER_API_KEY"] = SERPER_API_KEY
# print(SERPER_API_KEY)

# 3. Test Configuration
# url = "https://serper.dev"
url = "https://google.serper.dev/search"
payload = {"q": "Carl Friedrich Gauss major contributions"}
headers = {
    'X-API-KEY': SERPER_API_KEY,
    'Content-Type': 'application/json'
}

print("Checking Serper API connectivity...")

try:
    response = requests.post(url, headers=headers, json=payload)
    
    # This will catch 403 Forbidden errors immediately
    response.raise_for_status() 
    
    # If successful, print the results
    data = response.json()
    print("✅ Success! Your Serper API key is working perfectly.\n")
    print("--- Sample Search Results Found ---")
    for item in data.get('organic', [])[:3]:  # Print the top 3 headlines
        print(f"Title: {item.get('title')}")
        print(f"Link:  {item.get('link')}\n")

except requests.exceptions.HTTPError as e:
    print("❌ API Key Verification Failed!")
    print(f"Status Code: {e.response.status_code}")
    print(f"Error Message: {e.response.text}")
except Exception as e:
    print(f"An unexpected connection error occurred: {e}")
```
