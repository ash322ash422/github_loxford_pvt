```python
import os
from getpass import getpass
from dotenv import load_dotenv

# Load variables from the local .env file
load_dotenv()

# Check if the Gemini API key is already loaded in the environment
llm_api_key = os.getenv("GEMINI_API_KEY")
# llm_api_key = os.getenv("OPENAI_API_KEY")


if not llm_api_key:
    print("⚠️ GEMINI_API_KEY not found in environment variables.")
    # Prompt the user for the key securely without displaying the typed characters
    llm_api_key = getpass("🔐 Please enter your Gemini API Key: ")
    
    # Optional: Manually set it into the environment context for tracking libraries
    os.environ["GEMINI_API_KEY"] = llm_api_key

print("✅ Gemini API Key is ready for use.")
```

```python
# Check if the Gemini API key is already loaded in the environment
serper_api_key = os.getenv("SERPER_API_KEY")

if not serper_api_key:
    print("⚠️ SERPER_API_KEY not found in environment variables.")
    # Prompt the user for the key securely without displaying the typed characters
    serper_api_key = getpass("🔐 Please enter your Serper API Key: ")
    
    # Optional: Manually set it into the environment context for tracking libraries
    os.environ["SERPER_API_KEY"] = serper_api_key

print("✅ Serper API Key is ready for use.")
```

```python
from crewai import LLM

llm = LLM(
    # model="gemini/gemini-3.6-flash", # this model may get obsolete in the future
    model="openai/gpt-5-nano", # or use different model
)

response = llm.call("Give me a one line joke about fractals")

print(response)
```

```python
from crewai import Agent, Task, Crew
from crewai_tools import SerperDevTool

research_agent = Agent(
    role="Research Specialist",
    goal="Research interesting facts about the topic: {topic}",
    backstory="You are an expert at finding relevant and factual data.",
    tools=[SerperDevTool()],
    verbose=True,
    llm=llm
)

writer_agent = Agent(
    role="Creative Writer",
    goal="Write a short blog summary using the research",
    backstory="You are skilled at writing engaging summaries based on provided content.",
    llm=llm,
    verbose=True,
)
```

```python
task_research = Task(
    description="Find 3-5 interesting and recent facts about {topic} as of year 2025.",
    expected_output="A bullet list of 3-5 facts",
    agent=research_agent,
)

task_blog = Task(
    description="Write a 100-word blog post summary about {topic} using the facts from the research.",
    expected_output="A blog post summary",
    agent=writer_agent,
    context=[task_research],
)
```

```python
crew = Crew(
    agents=[research_agent, writer_agent],
    tasks=[task_research, task_blog],
    verbose=True,
)



# result = crew.kickoff(inputs={"topic": "Challenges of travelling to Moon"})
# print(result)

# In Jupyter, you can use 'await' directly at the cell level!
result = await crew.kickoff_async(inputs={"topic": "Challenges of travelling to Moon"})
print(result.raw)
```
